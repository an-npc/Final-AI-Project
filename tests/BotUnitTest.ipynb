```python
import sys, os
sys.path.append(os.path.abspath(".."))
import unittest
from models.bot.BaseBot import BaseBot
```

```python
class BotUnitTests(unittest.TestCase):
    """
    this class will only run methods with the phrase 'test_' in at the start of the method name
    """
    # walks through one betting round with 3 players, then the start of the next round

    def setUp(self):
        self.player_a = BaseBot()
        self.player_b = BaseBot()
        self.player_c = BaseBot()

        for player in (self.player_a, self.player_b, self.player_c):
            player.gain_chips(1000)

        self.pot = 0
        self.table_bet = 0


    # TESTING CALL METHOD
    def test_PlayersCall_Pot50_ReturnsNoErrors(self):
        self.pot = 50 # presummes that someone already bet 50
        self.pot += self.player_b.call(table_bet=self.pot)

        self.assertEqual(self.pot, 100)

        self.pot += self.player_c.call(table_bet=self.pot)

        self.assertEqual(self.pot, 200)



    def test_player_a_bets_50(self):
        # bet() isn't implemented yet, so call up to 50 to put the same chips in
        self.table_bet = 50
        self.pot += self.player_a.call(self.table_bet)

        self.assertEqual(self.table_bet, 50)
        self.assertEqual(self.pot, 50)
        self.assertEqual(self.player_a.get_chips(), 950)

    def test_calls_grow_pot_but_not_table_bet(self):
        self.table_bet = 50
        self.pot += self.player_a.call(self.table_bet)

        self.pot += self.player_b.call(self.table_bet)
        self.assertEqual(self.table_bet, 50)
        self.assertEqual(self.pot, 100)

        self.pot += self.player_c.call(self.table_bet)
        self.assertEqual(self.table_bet, 50)
        self.assertEqual(self.pot, 150)

        for player in (self.player_a, self.player_b, self.player_c):
            self.assertEqual(player.get_chips(), 950)
            self.assertEqual(player.get_current_bet(), 50)

    def test_next_round_resets_table_bet_keeps_pot(self):
        self.table_bet = 50
        for player in (self.player_a, self.player_b, self.player_c):
            self.pot += player.call(self.table_bet)

        # flop: new betting round
        self.table_bet = 0
        for player in (self.player_a, self.player_b, self.player_c):
            player.reset_roundt()

        self.assertEqual(self.table_bet, 0)
        self.assertEqual(self.pot, 150)
        for player in (self.player_a, self.player_b, self.player_c):
            self.assertEqual(player.get_current_bet(), 0)
```

```python
unittest.main(argv=[''], exit=False, verbosity=2)
```
